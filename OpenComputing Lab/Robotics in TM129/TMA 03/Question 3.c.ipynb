```python
%%sim_magic_preloaded -Cp -b Empty_Map -x 100 -y 700 -a -90

from ev3dev2.motor import MoveTank, OUTPUT_B, OUTPUT_C, SpeedPercent

# Initialize the MoveTank instance using standard ports
# OUTPUT_B is the left motor, OUTPUT_C is the right motor
tank = MoveTank(OUTPUT_B, OUTPUT_C)

# 1. Drive forward a short way
# Both motors run at 50% speed for 2 seconds to create the straight line
tank.on_for_seconds(SpeedPercent(50), SpeedPercent(50), 2)

# 2. Turn through approximately 360 degrees pivoting around one wheel
# To trace the circle BELOW the line (turning clockwise/right),
# the left motor drives forward while the right motor remains stationary (0%).
# The time value (e.g., 4.3 seconds) is an approximation to complete the full 360-degree pivot.
tank.on_for_seconds(SpeedPercent(50), SpeedPercent(0), 4.4)

# 3. Drive forward a short way
# Both motors run at 50% speed for 2 seconds to create the straight line
tank.on_for_seconds(SpeedPercent(50), SpeedPercent(50), 2)
```
