```python
%%sim_magic_preloaded -b Rainbow_bands --pendown --pencolor red --clear
import time
from ev3dev2.motor import MoveSteering, OUTPUT_B, OUTPUT_C
from ev3dev2.sensor.lego import ColorSensor
from ev3dev2.sensor import INPUT_2

# 1. Initialization
steering_drive = MoveSteering(OUTPUT_B, OUTPUT_C)
color_sensor = ColorSensor(INPUT_2)

# Start the robot moving forward 
steering_drive.on(0, 20)

# 2. First Encounter: Dark blue band (Robot sees 'Blue')
while color_sensor.color_name != 'Blue': 
    time.sleep(0.05)

say('turn here')
print('turn here')

# 3. Second Encounter: Boundary of Magenta and Yellow 
# First, wait until we enter the magenta band (Robot sees 'Red')
while color_sensor.color_name != 'Red':
    time.sleep(0.05)

# Next, wait until we enter the yellow band
while color_sensor.color_name != 'Yellow':
    time.sleep(0.05)

say('turn here')
print('turn here')

# 4. Third Encounter: End of the black band
# First, wait until we enter the black band
while color_sensor.color_name != 'Black':
    time.sleep(0.05)

# Next, wait until we are NO LONGER on the black band (hitting 'White')
while color_sensor.color_name == 'Black':
    time.sleep(0.05)

# 5. Stop and do half a turn on the spot
steering_drive.off()

# Use time instead of degrees to prevent the simulator from getting stuck.
# Steering = 100 (turn in place), Speed = 20%, Time = 2.5 seconds
steering_drive.on_for_seconds(100, 20, 3) 

# 6. Stop altogether
steering_drive.off()
```
