```python
import glob
import os

from PIL import Image, ImageDraw, ImageFont

image_filename = "SS9_car_305_top_40.png"

# 1. Download image if missing
if not os.path.exists(image_filename):
    print(f"'{image_filename}' not found locally. Downloading from OU repository...")
    url = f"https://raw.githubusercontent.com/innovationOUtside/TM129-23J/main/{image_filename}"
    try:
        urllib.request.urlretrieve(url, image_filename)
        print("Download successful!")
    except Exception:
        # Fallback to creating a blank canvas if offline or URL shifts
        print("Could not download file. Creating local canvas...")
        im = Image.new("RGB", (800, 600), color=(220, 220, 220))
        im.save(image_filename)

# 2. Load Image
im = Image.open(image_filename)
position = (100, 100)
text = "Explanatory text"
color = "black"

# 3. macOS Font Setup
try:
    font = ImageFont.truetype(
        "/System/Library/Fonts/Supplemental/Times New Roman Bold.ttf", 64
    )
except OSError:
    font = ImageFont.load_default(size=64)

# 4. Modern Pillow text dimension calculation (replacing font.getsize)
left, top, right, bottom = font.getbbox(text)
text_width = right - left
text_height = bottom - top

# 5. Draw overlay
draw = ImageDraw.Draw(im)
draw.rectangle(
    [position, (position[0] + text_width + 10, position[1] + text_height + 10)],
    fill="orange",
    outline="orange",
    width=5,
)
draw.text(position, text, fill=color, font=font)
draw.line([(0, 0), position], fill="red", width=5)

# Display in notebook
im
```

```python
# https://stackoverflow.com/a/52574134/454773
coords = None

%matplotlib notebook
import ipywidgets as wdg  # Using the ipython notebook widgets
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()
plt.imshow(im)

# Create and display textarea widget
txt = wdg.Textarea(value="", placeholder="", description="Co-ords:", disabled=False)
display(txt)


# Define a callback function that will update the textarea
def onclick(event):
    global coords
    coords = event
    txt.value = f"{coords.x}, {coords.y}"  # Dynamically update the text box above


# Create an hard reference to the callback not to be cleared by the garbage collector
ka = fig.canvas.mpl_connect("button_press_event", onclick)
```
