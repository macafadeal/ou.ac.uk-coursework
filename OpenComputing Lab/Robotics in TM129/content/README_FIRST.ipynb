```python
import datetime
import sys
from importlib.metadata import distributions

# 1. Format Python version and timestamp (DD.MM.YYYY-HHMMSS)
py_ver = sys.version.split()[0]
timestamp = datetime.datetime.now(datetime.timezone.utc).strftime("%d.%m.%Y-%H%M%S")

# 2. Construct target filename
filename = f"backup-requirements-python-{py_ver}-{timestamp}.txt"

# 3. Read installed packages directly from Python's package metadata
packages = []
for dist in sorted(distributions(), key=lambda d: d.name.lower()):
    name = dist.name
    version = dist.version
    packages.append(f"{name}=={version}")

# 4. Write to file
with open(filename, "w") as f:
    f.write("\n".join(packages) + "\n")

print(f"Successfully saved {len(packages)} packages to: {filename}")
```

```python
import os
import urllib.request

from PIL import Image, ImageDraw, ImageFont

image_filename = "SS9_car_305_top_40.png"

# 1. Handle missing image (Download or create placeholder)
if not os.path.exists(image_filename):
    print(
        f"Warning: {image_filename} not found locally. Creating a placeholder image..."
    )
    # Create a simple background canvas so your code can run uninterrupted
    im = Image.new("RGB", (800, 600), color=(240, 240, 240))
else:
    im = Image.open(image_filename)

# 2. Cross-platform Font Selection (macOS / Fallback)
font_size = 64
font = None

# macOS font paths to try
mac_fonts = [
    "/System/Library/Fonts/Supplemental/Georgia Bold.ttf",
    "/System/Library/Fonts/Supplemental/Times New Roman Bold.ttf",
    "/System/Library/Fonts/Helvetica.ttc",
    "/Library/Fonts/Arial.ttf",
]

for font_path in mac_fonts:
    if os.path.exists(font_path):
        try:
            font = ImageFont.truetype(font_path, font_size)
            break
        except Exception:
            continue

# Default back to Pillow's built-in scalable font if system TTFs aren't reached
if font is None:
    font = ImageFont.load_default(size=font_size)

# 3. Draw text on image
draw = ImageDraw.Draw(im)
position = (100, 100)
text = "Explanatory text"
draw.text(position, text, fill=(0, 0, 0), font=font)

# Display result in Jupyter
im
```
